# 测评流程

```python
# 一个问题
def evaluate_LLM_hallucination(client1,client2,client1_accurary,client2_accurary,question,anwser):# 裁判1(deepseek)、裁判2(gemini)、选手、答案

  response1 = client1.chat.completions.create(
    model="deepseek-chat",
    messages=[
        { "role": "system", "content": "我将会给你一个问题以及一个别的大语言模型给出的答案，请你判断这个回答是否出现幻觉现象，你只需要回答是或不是。" },
        { "role": "user", "content": f"问题为：{question}；答案为：{anwser}" }
    ],
    stream=False
  )
  response1 = response1.choices[0].message.content.strip()
  response2 = client2.models.generate_content(
      model="gemini-2.0-flash",
      contents=f'我将会给你一个问题以及一个别的大语言模型给出的答案，请你判断这个回答是否出现幻觉现象，你只需要回答是或不是。问题为：{question}；答案为：{anwser}',
  )
  response2 = response2.text.strip()

  # 结合正确率给分
  if response1 == "是" and response2 == "是":
      print("ds认为 是 ，gemini认为 是 ")
      return client1_accurary + client2_accurary -client2_accurary*client1_accurary,anwser
  elif response1 == "不是" and response2 == "是":
      print("ds认为 不是 ，gemini认为 是 ")
      return (1-client1_accurary)*client2_accurary,anwser
  elif response1 == "是" and response2 == "不是":
      print("ds认为 是 ，gemini认为 不是 ")
      return (1-client2_accurary)*client1_accurary,anwser
  elif response1 == "不是" and response2 == "不是":
      print("ds认为 不是 ，gemini认为 不是")
      return (1-client1_accurary)*(1-client2_accurary),anwser

```
